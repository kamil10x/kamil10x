```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

iris = load_iris()

x = iris.data
y = iris.target

model = DecisionTreeClassifier()
model.fit(x, y)

new = [[5.1, 4.2, 1.0, 0.5]]

z = model.predict(new)

print("Prediction: ", z[0])
print("Class: ", iris.target_names[z[0]])
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

x = np.array([[1], [2], [3], [4], [5]])
y = np.array([1, 4, 9, 16, 25])

model = LinearRegression()
model.fit(x, y)

z = model.predict(x)

print("Actual: ", y)
print("Prediction: ", z)
print("Mean Squared error: ", mean_squared_error(y, z))
```

```python
#lasso
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import Lasso
import numpy as np

x = np.array([[-4], [-3], [-2], [-1], [0], [1], [2], [3], [4]])
y = np.array([16, 9, 4, 1, 0, 1, 4, 9, 16])

model = Lasso(alpha=0.2)
model.fit(x, y)

z = model.predict(x)

print("Predicted: ", z)
print("Error: ", mean_squared_error(y, z))
```

```python
#Ridge
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import Ridge
import numpy as np

x = np.array([[-4], [-3], [-2], [-1], [0], [1], [2], [3], [4]])
y = np.array([-16, -9, -3, -1, 0, 1, 4, 9, 16])

model = Ridge(alpha=0.2)
model.fit(x, y)

new = model.predict(x)

print("Actual: ", y)
print("Predicted: ", new)
print("Error: ", mean_squared_error(y, new))
```

```python
#Adaboost
import numpy as np
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier

x = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

wmodel = DecisionTreeClassifier()

model = AdaBoostClassifier(estimator=wmodel, n_estimators=5, random_state=42, learning_rate=1)

model.fit(x, y)

yp = model.predict(x)

print("Actual: ", y)
print("Predicted: ", yp)

new = np.array([[4.5]])
pr = model.predict(new)

print("Prediction: ", pr[0])
```

```python
#GradientBoosting
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor

x = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])
y = np.array([10, 20, 30, 40, 50, 60, 70, 80])

model = GradientBoostingRegressor(
    n_estimators=9,
    random_state=42,
    learning_rate=0.2
)
model.fit(x, y)

yp = model.predict(x)

print("Actual: ", y)
print("Predicted: ", yp)

new = np.array([[4.5]])
z = model.predict(new)
print("Prediction: ", z)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree

iris = load_iris()

x = iris.data
y = iris.target

fnames = iris.feature_names
cnames = iris.target_names

model = DecisionTreeClassifier(criterion = "entropy", random_state=0, max_depth=2)
model.fit(x, y)

new = pd.DataFrame([[5.1, 3.4, 1.2, 0.2]], columns = fnames)
pr = model.predict(new)

print("Prediction: ", pr[0])
print("class: ", cnames[pr[0]])

plt.figure(figsize=(15, 20))
plot_tree(model, feature_names=fnames, class_names=cnames, filled= True)
plt.show()
```

```python
#Batch gradient descent
import numpy as np
import matplotlib.pyplot as plt

x = np.array([5, 10, 15, 20, 25])
y = np.array([15, 30, 45, 60, 75])

b0 = 0
b1 = 0

alpha = 0.001
itr = 100
n = len(x)

for i in range(itr):
    yp = b0 + b1 * x
    err = yp - y
    
    db0 = (1/n)*np.sum(err)
    db1 = (1/n)*np.sum(err * x)
    
    b0 = b0 - alpha*db0
    b1 = b1 - alpha*db1

cost = (1/(2*n))*np.sum(err**2)
print("Final prediction: ")
print(f"{b0: .2f} + {b1: .2f}x")

print("Cost: ",cost)
#yp = b0 + b1 * x
plt.plot(x, y, label = "Actual")
plt.plot(x, yp, label = "Predicted")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.legend()
plt.show()
```
